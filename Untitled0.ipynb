```python
import re
```

```python
#. - Entender qualquer valor exceto uma nova linha
#\. - busca o caractere "."
texto = 'o macaco é malaco'
padrao = re.compile(r'ma.aco')
check = padrao.findall(texto)
print(check)
```

```python
texto = '''
cachaça
cathaça
cabhaça
caxhaça
'''
padrao = re.compile(r'ca.haça')
check = padrao.findall(texto)
print(check)
```

```python
#^ - Vai olhar para o início da string
#[^] -Considerará todos os caracteres, exceto o indicado
texto = 'urubu'
padrao1 = re.compile(r'^u') #checa se o primeiro caractere é u
padrao2 = re.compile(r'[û]') #procura os caracteres diferentes de u

check1 = padrao1.findall(texto)
check2 = padrao2.findall(texto)

print(check1)
print(check2)
```

```python
#\d - Qualquer caractere que seja um algarismo de 0 a 9
#\D Qualquer caractere que NÃO SEJA um algarismo de 0 a 9
texto = 'urubu2009'
padrao1 = re.compile(r'\d') #uso do r 'raw string para evitar problema com a contrabarra, contrabarra é \n
padrao2 = re.compile(r'[\D]')

check1 = padrao1.findall(texto)
check2 = padrao2.findall(texto)

print(check1)
print(check2)
```

```python
#\s procura por caracteres vazios
#\S procura por qualquer caractere não vazio
texto = '''

urubu 2019

'''

padrao1 = re.compile(r'\s')
padrao2 = re. compile(r'\S')

check1 = padrao1.findall(texto)
check2 = padrao2.findall(texto)

print(check1)
print(check2)
```

```python
#\w - qualquer caractere que Seja alfanúmerico
#\W qualquer caractere que Não Seja alfanúmerico
texto = '''

urubu!_@ 2019-*%


'''
padrao1 = re.compile(r'\w')
padrao2 = re.compile(r'\W')

check1 = padrao1.findall(texto)
check2 = padrao2.findall(texto)
print(check1)
print(check2)
```

```python
texto = '''
Flamengo 1992

'''
pattern = re.compile(r'a-zA-Z0-9')

validacoes = pattern.finditer(texto)

for validacao in validacoes:
  print(validacao)


```

```python
texto = '''
Flamengo 1992
Baralho 2025
Noite5896
'''

pattern = re.compile(r'[a-zA-Z]+ [0-9]+')

validacoes = pattern.finditer(texto)

for validacao in validacoes:
  print(validacao) #.group() pega só o valor do match
```
